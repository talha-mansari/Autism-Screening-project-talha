```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
autistic_spectrum_disorder_screening_data_for_children = fetch_ucirepo(id=419) 
  
# data (as pandas dataframes) 
X = autistic_spectrum_disorder_screening_data_for_children.data.features 
y = autistic_spectrum_disorder_screening_data_for_children.data.targets 
  
# metadata 
print(autistic_spectrum_disorder_screening_data_for_children.metadata) 
  
# variable information 
print(autistic_spectrum_disorder_screening_data_for_children.variables) 
```

The above cell confirms my suspicions that the diagnosis is based on a Boolean expression from the scores of the test and not influenced by the other factors. Let's drop the results column and see how the model fairs:

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
```

```python
for col in df.select_dtypes('object').columns:
    df[col] = df[col].str.strip("' ")
```

```python
cat_cols = ['gender', 'ethnicity', 'jaundice', 'autism',
            'country_of_res', 'used_app_before', 'relation']

ohe = OneHotEncoder(handle_unknown='infrequent_if_exist', min_frequency=5,
                    sparse_output=False).set_output(transform="pandas")

encoded = ohe.fit_transform(df[cat_cols])
df_encoded = pd.concat([df.drop(columns=cat_cols), encoded], axis=1)
df_encoded.head()
```

```python
df_encoded.columns
```

```python
X = df_encoded.drop(columns=['result', 'age_desc', 'class'])
y = df_encoded['class']
```

```python
X.columns
```

```python
train_X, val_X, train_y, val_y = train_test_split(X, y,random_state = 0)
```

```python
model = RandomForestClassifier()
model.fit(train_X, train_y)
predictions = model.predict(val_X)


print(accuracy_score(val_y, predictions))
print(confusion_matrix(val_y, predictions, labels=['YES', 'NO']))
```
